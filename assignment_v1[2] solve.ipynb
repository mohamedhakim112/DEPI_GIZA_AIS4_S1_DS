# Create a function that Calculates the Sum of Even Numbers in a List
## Example
numbers: [1, 2, 3, 4, 5, 6]

Output: 12

```python
def sum_of_even(numbers):
    return sum(n for n in numbers if n % 2 == 0)

nums = [1, 2, 3, 4, 5, 6]
print(sum_of_even(nums))
```
